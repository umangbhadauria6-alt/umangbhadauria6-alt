```python
import kagglehub
path = kagglehub.dataset_download("shivamb/netflix-shows")
print("Path to dataset files:", path)
```

```python
import os
files = os.listdir(path)
print(files)
```

```python
import pandas as pd
df = pd.read_csv(path + "/netflix_titles.csv")
df.head()
```

```python
import matplotlib.pyplot as plt

type_counts = df['type'].value_counts()
type_counts.plot(kind='bar', color=['red', 'black'])
plt.title('Movies vs TV Shows on Netflix')
plt.xlabel('Type')
plt.ylabel('Count')
plt.show()
```

```python
country_counts = df['country'].value_counts().head(10)
country_counts.plot(kind='barh', color='darkred')
plt.title('Top 10 Countries by Netflix Content')
plt.xlabel('Number of Titles')
plt.show()
```

```python
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
df['year_added'] = df['date_added'].dt.year
yearly_counts = df['year_added'].value_counts().sort_index()
yearly_counts.plot(kind='line', marker='o', color='green')
plt.title('Netflix Content Added Per Year')
plt.xlabel('Year')
plt.ylabel('Number of Titles Added')
plt.show()
```

## Key Insights
- Movies make up the majority of Netflix's content compared to TV Shows.
- The United States leads in content production, followed by India and the UK.
- Netflix significantly increased the amount of content added between 2018-2020, likely reflecting global expansion.
